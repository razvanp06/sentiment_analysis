```python
import joblib
from preprocess import load_nlp, predict

nlp = load_nlp()
vectorizer = joblib.load('../models/vectorizer.joblib')
model = joblib.load('../models/model.joblib')

reviews = [
	'Foarte mulțumit, se încarcă repede și ține mult.',
	'Nu îl recomand, s-a stricat după o săptămână.',
	'Nu e rău deloc pentru prețul ăsta.',
	'Arată bine, dar bateria e slabă.',
]
for r in reviews:
	print(predict(r, nlp, vectorizer, model), r)
```
